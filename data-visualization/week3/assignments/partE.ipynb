## Item 1 — Per-Panel Justification (6 marks)

*For each panel: the question it answers, why that chart type, which Week 2 channel rule it follows.*

### Panel 1 (top-left): Histogram of systolic blood pressure

**Question it answers:** Is systolic blood pressure normally distributed in this cohort, and how many patients are in the hypertensive range (≥130 mmHg)?

**Why a histogram:** This is the appropriate chart for showing the shape of a single continuous distribution. A box plot would show the median and IQR but would obscure the bimodal tendency and the long hypertensive tail that are clinically relevant. A KDE would impose a bandwidth-smoothing assumption that needs to be declared and justified; a histogram with a chosen bin count is more transparent.

**Week 2 channel rule:** The single quantity (count of patients per bin) is encoded on the **y-position** axis — the highest-ranking channel for quantitative data (Cleveland & McGill). The horizontal reference line at 130 mmHg uses **colour** (the course highlight colour `#D55E00`) to pop out the clinical threshold, which is the one element worth a reader's attention. All other ink is in the muted series colour.

---

### Panel 2 (top-right): Box plot of length of stay by ward

**Question it answers:** Which ward has the longest patient stays, and does the distribution differ systematically across wards?

**Why a box plot:** Length-of-stay is heavily right-skewed (most patients leave in 1–5 days; a small number have stays of 30+ days). A bar chart of means would be misleading — the mean is pulled by the long tail and does not represent a typical patient. A box plot shows the median, the middle 50%, and the outlier spread simultaneously. Wards are sorted by median to make comparison immediate.

**Log scale:** A linear y-axis collapses the entire interquartile range of General Medicine and Surgery into a few pixels because ICU outliers extend to 60+ days. A log scale preserves the readable comparison of medians and IQRs across all four wards.

**Fliers shown:** Box plot outliers are shown (`showfliers=True`). Hiding them with `showfliers=False` would suppress the very data anomalies this clinical review was commissioned to find. The source note on the figure records this decision.

**Week 2 channel rule:** Category identity (ward) is encoded on the **x-position** axis (nominal axis). The measured value (LOS days) is encoded on the **y-position** axis. Colour supplements identity but is not necessary — each ward is already identified by its x-label. The palette uses the three course series colours plus the muted colour for the fourth ward, with direct x-axis labelling so colour is redundant, not necessary.

---

### Panel 3 (bottom-left): Scatter of systolic vs diastolic blood pressure

**Question it answers:** How strongly are systolic and diastolic blood pressure correlated in this cohort, and does the relationship appear linear?

**Why this pair (from the heatmap):** The heatmap in panel 4 was built first, following the task instruction. It shows that systolic and diastolic have the strongest clinically meaningful correlation that is not arithmetically forced. (`map_mmhg` was excluded because it is computed from those two values — including it would be reporting an arithmetic identity as a finding.)

**Why a scatter plot:** A scatter is the correct chart for showing the relationship between two continuous variables. A line chart would be wrong (no ordered x-axis in the observational sense). A heatmap would aggregate the relationship and lose individual-point distribution.

**Subsample:** 10,000 overlapping dots produce a solid black blob. A subsample of 2,000 rows (seed=42) is shown, with `alpha=0.25` to reveal density through opacity. The full n is used for the regression line. The source note records the subsample size.

**Week 2 channel rule:** Both variables are quantitative. Both are encoded on **position** axes — the highest-ranking channel. No other channel is used for a third variable. The regression line uses the highlight colour to draw attention without adding a category.

---

### Panel 4 (bottom-right): Annotated correlation heatmap

**Question it answers:** Which pairs of vital-sign measurements are correlated strongly enough to be clinically worth investigating further?

**Why a heatmap:** A heatmap is the correct screening tool for a correlation matrix. It allows the reader to locate both strong positive and strong negative correlations across all pairs at once, without having to scan a table of numbers. The heatmap is a *screening tool* (this is what Lecture 6 said it is for), not a final result — it directs the researcher to the scatter in panel 3.

**Chart construction rules followed:**
- `cmap='RdBu_r'` — a diverging palette centred at zero, because a correlation has a sign (positive and negative are different things, not just more or less of the same thing)
- `vmin=-1, vmax=1, center=0` — the scale is fixed at the true range of the Pearson r statistic; no autoscaling that would make weak correlations look strong
- Lower triangle only (`mask` the upper triangle) — the matrix is symmetric; showing both triangles says everything twice and doubles the ink without adding information
- `annot=True, fmt='.2f'` — numerical annotation makes the chart readable without a colour bar look-up; every cell declares its value directly
- `map_mmhg` excluded — see item 3 below

## Item 2 — Every Parameter Declared and Defended (5 marks)

| Parameter | Value | Reason |
|---|---|---|
| `bins` (histogram) | Freedman-Diaconis rule: bins = (max−min) / (2 × IQR × n^(−1/3)), rounded into [20, 60] | FD rule chooses bin width based on data spread and sample size, outperforming equal-width bins for skewed distributions (Hyndman, 1995). Hard minimum of 20 prevents over-aggregation; cap of 60 prevents noise bars. |
| `yscale='log'` (box plot) | Log | LOS days are log-normally distributed; a linear scale collapses the IQR of 3 of the 4 wards to a narrow band because ICU outliers extend to 60+ days. |
| `showfliers=True` | True | This is a *clinical review*. Hiding outliers would conceal the anomalies the review was commissioned to find. |
| `ward_order` (box plot) | Sorted by median ascending | Sorted ordering makes the comparison of medians across wards immediate. Alphabetical ordering carries no information about the data. |
| `alpha=0.25` (scatter) | 0.25 | At 2,000 points the scatter is still dense. Alpha 0.25 lets overlapping points accumulate as darker regions, showing density without overplotting to a solid blob. |
| Subsample size (scatter) | 2,000 (seed=42) | 10,000 overlapping points become a single opaque mass even at low alpha values. 2,000 shows the relationship shape and density gradient clearly. Seed is fixed so the figure is reproducible. |
| `cmap='RdBu_r'` (heatmap) | RdBu_r | Diverging palette for a signed statistic. Red = positive correlation, Blue = negative. `_r` reverses so red (warm, positive) is conventional. The course palette specifies `diverging = 'RdBu_r'` for exactly this case. |
| `vmin=-1, vmax=1` (heatmap) | −1 to +1 | The Pearson r statistic is bounded at ±1 by definition. Autoscaling would make weak correlations (r=0.3) look as intense as strong ones (r=0.9). Fixed limits make colour intensity mean something. |
| `center=0` (heatmap) | 0 | Centres the diverging colour scale at the neutral point (no correlation). Without it, the colour ramp is off-centre and positive/negative correlations are not visually symmetric. |
| `mask` (heatmap) | Upper triangle | Mask = `np.triu(..., k=0)` (including the diagonal). Showing only the lower triangle removes the redundant mirror image. The diagonal (r=1 with itself) is also masked — it adds no information. |
| `annot=True, fmt='.2f'` (heatmap) | Two decimal places | The correlation coefficient is reported to 2 decimal places — the precision that is meaningful (r=0.77 is meaningfully different from r=0.7; r=0.773 vs r=0.771 is not). |
| `dpi=300` (save) | 300 | The TASKS.md rule and course slide 27: `fig.savefig(dpi=300, bbox_inches='tight')`. Print quality requires ≥300 dpi. |
| `bbox_inches='tight'` (save) | tight | Prevents the source note (added via `fig.text`) from being clipped at the bottom of the canvas. |
| BMI sentinel threshold | >200 | The maximum physiologically plausible BMI in any recorded case is well below 200 kg/m². The value 999 is a common sentinel for 'missing' in clinical data entry systems. |
| Heart-rate fault threshold | >250 bpm | 250 bpm is above the maximum sustainable human heart rate (220 − age, with the lowest ages in the cohort giving ~190 bpm theoretical max). Readings at 300–400 bpm are device artefacts. |

## Item 3 — Every Row Removed (4 marks)

*Rule, count, and what the statistic moved from and to.*

```python
import os, sys, pandas as pd, numpy as np, warnings
warnings.filterwarnings('ignore')
sys.path.insert(0, os.path.join(os.getcwd(), '..'))
from vizlib import PALETTE

raw = pd.read_csv(os.path.join('..', 'data', 'vitals_10k.csv'))

# 1. BMI sentinels
bmi_before = raw['bmi'].mean()
n_bmi = (raw['bmi'] > 200).sum()
raw_bmi = raw.copy()
raw_bmi.loc[raw_bmi['bmi'] > 200, 'bmi'] = np.nan
bmi_after = raw_bmi['bmi'].mean()

# 2. Heart-rate device faults
hr_before = raw['heart_rate'].mean()
n_hr = (raw['heart_rate'] > 250).sum()
hr_after = raw.loc[raw['heart_rate'] <= 250, 'heart_rate'].mean()

print('Exclusion report:')
print(f'  Total rows loaded:                        {len(raw):,}')
print()
print(f'  BMI sentinels (bmi > 200 → np.nan):')
print(f'    Rows affected (column-level, not row drop): {n_bmi:,}')
print(f'    Mean BMI before: {bmi_before:.2f}  after: {bmi_after:.2f}')
print()
print(f'  Heart-rate device faults (heart_rate > 250 bpm → row removed):')
print(f'    Rows removed: {n_hr:,}')
print(f'    Mean HR before: {hr_before:.1f}  after: {hr_after:.1f} bpm')
print()
print(f'  Rows in final dataset: {len(raw) - n_hr:,}')
```

### Narrative Summary

**Exclusion 1 — BMI sentinel values (column-level replacement, not row removal)**

The `bmi` column contains the value 999.0 as a sentinel for 'patient record incomplete'. This is a standard practice in some clinical data entry systems. These values are replaced with `np.nan` (not removed as rows, because the other measurements for those patients are valid).

- Rows affected: see code output above
- Mean BMI moves from the inflated value (including 999) to the physiologically plausible range
- **Symmetry test:** I would make this replacement regardless of whether it strengthened or weakened any finding. A BMI of 999 kg/m² does not exist. There is no alternative reading of the value.

**Exclusion 2 — Heart-rate device faults (row removal)**

`heart_rate` readings above 250 bpm are device artefacts. No human heart sustains a rate of 300–400 bpm. The physiological maximum for any person in this cohort (youngest = age ~18, max theoretical HR ≈ 202 bpm) is well below 250.

- Rows removed: see code output above
- Mean HR moves from the inflated value to a physiologically plausible range
- **Ethics clause (TASKS.md):** These rows are removed *and stated*. Using `showfliers=False` in panel 2 without this disclosure would have hidden device faults from the clinical reviewer without their knowledge — which is the scenario the ethics clause in TASKS.md describes. The source note on the figure carries the rule and the count.
- **Symmetry test:** I would make this removal regardless of direction. A 400 bpm reading is a device fault in any clinical context. If the faults had *reduced* the apparent mean HR instead of inflating it, the removal would still be correct.

**Derived column excluded from heatmap — not a row removal**

`map_mmhg` (mean arterial pressure) is computed as `diastolic + (systolic − diastolic) / 3`. Its correlation with `systolic` and `diastolic` is arithmetic, not a clinical finding. Including it in the heatmap would display a near-perfect correlation (r ≈ 0.96 with systolic) that carries no new information and could mislead a reviewer into thinking they have discovered a strong relationship. It is excluded from panel 4 and the source note records the reason.

**What was NOT done:**

- `spo2` readings of 100% are not removed. 100% is a real physiological value (fully saturated). The KDE boundary problem (Part A, Case 9) occurs because the kernel estimate extends above 100% — the fix is to use a bounded KDE or constrain the x-axis, not to remove valid data.
- No records were removed for being 'outliers' without a clinical justification. A 30-day hospital stay is clinically plausible for an ICU patient; it is not an error.
