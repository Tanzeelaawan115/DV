**Goal:** Show that summary statistics can be identical across wildly different shapes.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('data/datasaurus.csv')
print(f'Shape: {df.shape}')
print(f'Unique datasets: {df["dataset"].nunique()}')
print(f'Points per dataset: {df.groupby("dataset").size().unique()}')
df.head()
```

## Step 1 — Summary Statistics for Each of the 13 Datasets

```python
stats = df.groupby('dataset').apply(
    lambda g: pd.Series({
        'mean_x':  round(g['x'].mean(), 2),
        'mean_y':  round(g['y'].mean(), 2),
        'std_x':   round(g['x'].std(), 2),
        'std_y':   round(g['y'].std(), 2),
        'corr_xy': round(g['x'].corr(g['y']), 4)
    })
)
print(stats.to_string())
stats
```

## Step 2 — Conclusion from the Table Alone

Looking at the table, every dataset has nearly identical summary statistics:
- Mean of x ≈ 54.26, Mean of y ≈ 47.83
- Std of x ≈ 16.77, Std of y ≈ 26.94
- Correlation ≈ −0.0645

If the table were all we had, we would conclude that all 13 datasets are essentially the same — same location, same spread, same (near-zero) linear association. We would have no reason to treat them differently from each other.

## Step 3 — Grid of 13 Scatter Plots

```python
datasets = sorted(df['dataset'].unique())
ncols = 4
nrows = -(-len(datasets) // ncols)  # ceiling division

fig, axes = plt.subplots(nrows, ncols, figsize=(16, 13), constrained_layout=True)
flat = axes.flatten()

for i, ds in enumerate(datasets):
    sub = df[df['dataset'] == ds]
    flat[i].scatter(sub['x'], sub['y'], s=14, alpha=0.75, color='#2166ac', edgecolors='none')
    flat[i].set_title(ds, fontsize=10, fontweight='bold')
    flat[i].set_xlim(-5, 105)
    flat[i].set_ylim(-5, 105)
    for sp in ('top', 'right'):
        flat[i].spines[sp].set_visible(False)

for j in range(i + 1, len(flat)):
    flat[j].set_visible(False)

fig.suptitle('Identical statistics, completely different shapes — always plot your data',
             fontsize=13, fontweight='bold')
plt.savefig('task1_scatter_grid.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Step 4 — What We Actually Found

The 13 datasets span a remarkable variety of shapes: a dinosaur, a star, an X, a bullseye, a high scatter, a wide scatter, parallel lines, a circle, and several others. Yet their means, standard deviations, and correlations agree to two decimal places across all 13.

This is the strongest possible demonstration that **summary statistics do not describe shape**. They faithfully summarise location, spread, and linear association — but are completely silent about the underlying geometry.

## Answer: Does Having 13 Datasets and 142 Points Make Statistics More Trustworthy?

**No.** More datasets and more points do not make summary statistics a trustworthy substitute for plotting. The additional datasets multiply false confidence — each one separately confirms the same numbers — but they all still hide the same fundamental problem. Having 142 points instead of Anscombe's 11 only means that small-sample variation cannot be blamed for the result. The statistics are *exact*, and they are still lying.

The lesson: visualisation is not something you do after the statistics pass inspection. It is the step you do *before* you trust them.
